```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
from numpy import exp

# ===============================
# Daten einlesen
# ===============================
df = pd.read_excel("Exp2-3_p(x)_T(x)_Template2025Kopie.xlsx", engine="openpyxl",
                   usecols="A,H,I,J", skiprows=6, nrows=9)
df.columns = ["x_A", "y_B", "p_ges_theo", "p_ges"]
df["x_B"] = 1 - df["x_A"]

# ===============================
# Fitfunktion für x_B (Flüssigkeit)
# ===============================
x = df["x_B"].to_numpy()
p_ges = df["p_ges"].to_numpy()
pA0 = df["p_ges"].iloc[-1]
pB0 = df["p_ges"].iloc[0]

R = 8.314
T = 298.15

def p_ges_theorie(x, Omega):
    return x * pB0 * exp(Omega / (R * T) * (1 - x)**2) + (1 - x) * pA0 * exp(Omega / (R * T) * x**2)

# Fit durchführen
popt, _ = curve_fit(p_ges_theorie, x, p_ges, p0=[-100])
Omega_fit = popt[0]
print(f"Gefittetes Ω: {Omega_fit:.2f} J/mol")

# Glatte rote Kurve berechnen
x_fit = np.linspace(0, 1, 300)
y_fit = p_ges_theorie(x_fit, Omega_fit)

# ===============================
# Polynomfit für y_B (Gasphase)
# ===============================
sorted_df = df.sort_values("y_B")
y_B_sorted = sorted_df["y_B"].to_numpy()
p_sorted = sorted_df["p_ges"].to_numpy()

coeffs_blue = np.polyfit(y_B_sorted, p_sorted, deg=2)
poly_blue = np.poly1d(coeffs_blue)
y_B_smooth = np.linspace(y_B_sorted.min(), y_B_sorted.max(), 300)
p_blue_smooth = poly_blue(y_B_smooth)

# ===============================
# Fit für theoretische p_ges (z. B. aus Spalte I)
# ===============================
p_ges_theo = df["p_ges_theo"].to_numpy()

# Fit durch p_ges_theo (gleiche Formel wie oben)
popt_theo, _ = curve_fit(p_ges_theorie, x, p_ges_theo, p0=[-100])
Omega_theo = popt_theo[0]
print(f"Theorie-Fit: Ω = {Omega_theo:.2f} J/mol")

# Fit-Kurve berechnen
y_fit_theo = p_ges_theorie(x_fit, Omega_theo)

# ===============================
# Plot erstellen
# ===============================
plt.figure(figsize=(8, 6))

# Rote Fit-Kurve
plt.plot(x_fit, y_fit, color='red', label=f"x_B (Flüssigkeit, Fit mit Ω={Omega_fit:.1f} J/mol)", linewidth=2)

# Blaue Fit-Kurve
plt.plot(y_B_smooth, p_blue_smooth, color='blue', linewidth=2, label="y_B (Gasphase, Fit)")

# Theorie-Fit-Kurve (z. B. p_ges_theo)
plt.plot(x_fit, y_fit_theo, color='orange', linewidth=2, linestyle='--',
         label=f"x_B (Theorie-Fit mit Ω={Omega_theo:.1f} J/mol)")

# Messpunkte
plt.scatter(df["x_B"], df["p_ges"], color='red', label="Messpunkte x_B", marker='o')
plt.scatter(df["y_B"], df["p_ges"], color='blue', label="Messpunkte y_B", marker='o')

# Theoretische p_ges-Werte (z. B. aus Spalte I)
plt.scatter(df["x_B"], df["p_ges_theo"], color='orange', label="Theoriepunkte x_B", marker='x')

# Fit für Theoriepunkte y_B vs p_ges_theo
sorted_theo_yB = df.sort_values("y_B")
y_B_theo_sorted = sorted_theo_yB["y_B"].to_numpy()
p_theo_sorted = sorted_theo_yB["p_ges_theo"].to_numpy()

# Polynomfit
coeffs_theo_yB = np.polyfit(y_B_theo_sorted, p_theo_sorted, deg=2)
poly_theo_yB = np.poly1d(coeffs_theo_yB)

y_B_theo_fit = np.linspace(y_B_theo_sorted.min(), y_B_theo_sorted.max(), 300)
p_theo_yB_fit = poly_theo_yB(y_B_theo_fit)
plt.scatter(df["y_B"], df["p_ges_theo"], color='orange', label="Theoriepunkte y_B", marker='+')

# Verbindungslinien
for i in range(len(df)):
    plt.plot([df["x_B"][i], df["y_B"][i]], [df["p_ges"][i]] * 2, color='gray', linestyle='--', alpha=0.4)

# Zusätzliche Linien für p_ges_theo
for i in range(len(df)):
    plt.hlines(df["p_ges_theo"][i], 0, 1, colors='orange', linestyles='dotted', linewidth=1.2, alpha=0.6)

# Layout
plt.xlabel("Molenbruch x_B / y_B")
plt.ylabel("Dampfdruck in mbar")
plt.title("Zustandsdiagramm p(x_B), p(y_B)", fontsize=14, fontweight="bold")
plt.grid(True)
plt.legend()
plt.xlim(0, 1)
plt.ylim(0, max(df["p_ges"].max(), df["p_ges_theo"].max()) * 1.1)
plt.tight_layout()
plt.show()
```
